```python
def average(*numbers: list[float]):
    """Renvoie la moyenne des valeurs passées en paramètre"""
    return sum(numbers) / len(numbers)
```

```python
print(average(1, 2, 1.5, 3, 66, 17))

nbs = [1,2,3]
print(average(*nbs))
```

```python
"""
F0 = 0
F1 = 1
F2 = 1
F3 = 2
F4 = 3
F5 = 5

FN = FN-1 + FN-2

F34 = F33 + F32
"""
from functools import cache

@cache
def fibo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibo(n-1) + fibo(n-2)

print(fibo(100))
```

```python
def filtrerPair(numbers: list[int]):
    return [nb for nb in numbers if nb % 2 == 0]

def filtrerImpair(numbers: list[int]):
    return [nb for nb in numbers if nb % 2 != 0]

def filtrerPlusGrandQue50(numbers: list[int]):
    return [nb for nb in numbers if nb > 50]

def filtrer(numbers: list[int], cond):
    return [nb for nb in numbers if cond(nb)]


nbs = [99,25,66,33]

print(filtrer(nbs, lambda nb: nb % 3 == 0))



nbs = [1,2,3,4,5,6]
print(list(map(lambda n: n**2, nbs)))
print([n**2 for n in nbs])
```

```python
import json

import requests

url = 'https://openholidaysapi.org/PublicHolidays'

response = requests.get(url=url, params={
    'countryIsoCode': 'BE',
    'languageIsoCode': 'FR',
    'validFrom': '2026-01-01',
    'validTo': '2026-12-31'
})

data = json.loads(response.content)
for j in data:
    print(j['name'][0]['text'], j['startDate'])
```
